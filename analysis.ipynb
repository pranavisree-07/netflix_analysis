```python
import pandas as pd

# Load the dataset we created
df = pd.read_csv("data/movies.csv")

# Print the table to the screen
df
```

```python
# 1. Sort the dataset to find the highest-rated shows first
print("--- TOP RATED SHOWS ---")
top_rated = df.sort_values(by="Rating", ascending=False)
print(top_rated[["Title", "Rating"]])

print("\n--------------------------------")

# 2. Calculate basic statistics for ratings and show duration
print("--- SUMMARY STATISTICS ---")
avg_rating = df["Rating"].mean()
avg_duration = df["DurationMin"].mean()

print(f"Average Show Rating: {avg_rating:.2f} / 10")
print(f"Average Episode Duration: {avg_duration:.1f} minutes")
```

```python
import matplotlib.pyplot as plt

# Set up the visual look of the chart
plt.figure(figsize=(8, 4))

# Create a bar chart showing the ratings of each show
plt.bar(df["Title"], df["Rating"], color="royalblue")

# Add chart labels and formatting
plt.title("Netflix Show Ratings Comparison", fontsize=14, fontweight="bold")
plt.xlabel("Show Title")
plt.ylabel("Rating (out of 10)")
plt.ylim(0, 10) # Set scale from 0 to 10
plt.xticks(rotation=30) # Tilt names so they don't overlap

# Clean up layout and display the chart natively on screen
plt.tight_layout()
plt.show()
```
